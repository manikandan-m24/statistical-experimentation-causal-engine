```python
import math
import statistics
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# ==============================================================================
# 1. SYNTHETIC EXPERIMENT & PRE-EXPERIMENT TELEMETRY SIMULATION (N = 12,000)
# ==============================================================================
np.random.seed(42)
n_users = 12000

df = pd.DataFrame({
    'user_id': [f'USR_{i:05d}' for i in range(1, n_users + 1)],
    'group': np.random.choice(['Control', 'Variant_B'], size=n_users, p=[0.5, 0.5]),
    'device': np.random.choice(['Desktop', 'Mobile', 'Tablet'], size=n_users, p=[0.55, 0.35, 0.10]),
    'plan_tier': np.random.choice(['Starter', 'Pro', 'Enterprise'], size=n_users, p=[0.60, 0.30, 0.10]),
    # Pre-experiment covariate (prior 14-day engagement score) for CUPED variance reduction
    'pre_exp_engagement': np.random.gamma(shape=3.0, scale=20.0, size=n_users).round(2)
})

# Post-experiment session duration correlated with pre-experiment engagement
noise = np.random.normal(loc=0, scale=35.0, size=n_users)
treatment_session_lift = np.where(df['group'] == 'Variant_B', 18.0, 0.0)
df['session_duration_sec'] = np.clip(
    4.2 * df['pre_exp_engagement'] + treatment_session_lift + noise, 15.0, None
).round(1)

# Conversion probabilities: Control ~12.0%, Variant_B ~14.1% (+2.1 pp absolute lift)
base_prob = np.where(df['group'] == 'Control', 0.120, 0.141)
device_adj = np.where(df['device'] == 'Desktop', 0.015, np.where(df['device'] == 'Mobile', -0.015, 0.0))
conv_prob = np.clip(base_prob + device_adj, 0.01, 0.95)
df['converted'] = np.random.binomial(1, conv_prob)

# Right-skewed Log-Normal revenue for converted users
rev_mean = np.where(df['group'] == 'Control', 4.10, 4.22)
df['revenue_usd'] = np.where(
    df['converted'] == 1,
    np.random.lognormal(mean=rev_mean, sigma=0.55, size=n_users).round(2),
    0.0
)

df.to_csv('statistical_experiment_data.csv', index=False)

# ==============================================================================
# 2. FIRST-PRINCIPLES STATISTICAL & CAUSAL INFERENCE ENGINE
# ==============================================================================
def normal_cdf(z):
    """Exact standard normal CDF via Gauss error function."""
    return 0.5 * (1.0 + math.erf(z / math.sqrt(2.0)))

def wilson_ci(successes, n, z=1.96):
    """Wilson score interval for binomial proportions."""
    p = successes / n
    denom = 1.0 + (z**2) / n
    center = (p + (z**2) / (2.0 * n)) / denom
    half_width = (z * math.sqrt((p * (1.0 - p) + (z**2) / (4.0 * n)) / n)) / denom
    return center - half_width, center + half_width

ctrl = df[df['group'] == 'Control']
var = df[df['group'] == 'Variant_B']
n_ctrl, n_var = len(ctrl), len(var)

# --- A. Sample Ratio Mismatch (SRM) Guardrail Check (Chi-Square GoF, df=1) ---
expected_n = n_users / 2.0
srm_chi2 = ((n_ctrl - expected_n)**2 / expected_n) + ((n_var - expected_n)**2 / expected_n)
# For df=1, Chi-Square survival function equals 2 * (1 - normal_cdf(sqrt(chi2)))
p_val_srm = 2.0 * (1.0 - normal_cdf(math.sqrt(srm_chi2)))

# --- B. Two-Proportion Z-Test & Cohen's h Effect Size ---
c_ctrl, c_var = int(ctrl['converted'].sum()), int(var['converted'].sum())
p_ctrl, p_var = c_ctrl / n_ctrl, c_var / n_var
p_pool = (c_ctrl + c_var) / (n_ctrl + n_var)

se_pool = math.sqrt(p_pool * (1.0 - p_pool) * (1.0 / n_ctrl + 1.0 / n_var))
z_conv = (p_var - p_ctrl) / se_pool
p_val_conv = 2.0 * (1.0 - normal_cdf(abs(z_conv)))
cohens_h = 2.0 * math.asin(math.sqrt(p_var)) - 2.0 * math.asin(math.sqrt(p_ctrl))

ci_ctrl = wilson_ci(c_ctrl, n_ctrl)
ci_var = wilson_ci(c_var, n_var)

# --- C. CUPED Variance Reduction on Session Duration ---
# Y_cuped = Y - theta * (X - E[X]), where theta = Cov(Y, X) / Var(X)
y = df['session_duration_sec'].to_numpy()
x = df['pre_exp_engagement'].to_numpy()
theta = np.cov(y, x, ddof=1)[0, 1] / np.var(x, ddof=1)
df['session_cuped'] = y - theta * (x - np.mean(x))

raw_ctrl = df[df['group'] == 'Control']['session_duration_sec'].to_numpy()
raw_var = df[df['group'] == 'Variant_B']['session_duration_sec'].to_numpy()
cuped_ctrl = df[df['group'] == 'Control']['session_cuped'].to_numpy()
cuped_var = df[df['group'] == 'Variant_B']['session_cuped'].to_numpy()

se_raw = math.sqrt(np.var(raw_var, ddof=1)/n_var + np.var(raw_ctrl, ddof=1)/n_ctrl)
se_cuped = math.sqrt(np.var(cuped_var, ddof=1)/n_var + np.var(cuped_ctrl, ddof=1)/n_ctrl)
var_reduction_pct = (1.0 - (se_cuped**2 / se_raw**2)) * 100.0
z_cuped = (np.mean(cuped_var) - np.mean(cuped_ctrl)) / se_cuped
p_val_cuped = 2.0 * (1.0 - normal_cdf(abs(z_cuped)))

# --- D. Tie-Corrected Mann-Whitney U Test (Skewed Payer Revenue) ---
payers_var = var[var['converted'] == 1]['revenue_usd'].to_numpy()
payers_ctrl = ctrl[ctrl['converted'] == 1]['revenue_usd'].to_numpy()
n1, n2 = len(payers_var), len(payers_ctrl)

combined_rev = np.concatenate([payers_var, payers_ctrl])
ranks = pd.Series(combined_rev).rank(method='average').to_numpy()
u1 = np.sum(ranks[:n1]) - (n1 * (n1 + 1)) / 2.0
u2 = n1 * n2 - u1
u_stat = min(u1, u2)

_, tie_counts = np.unique(combined_rev, return_counts=True)
tie_term = np.sum(tie_counts**3 - tie_counts) / ((n1 + n2) * (n1 + n2 - 1))
mu_u = (n1 * n2) / 2.0
sigma_u = math.sqrt((n1 * n2 / 12.0) * ((n1 + n2 + 1) - tie_term))
z_u = (u_stat - mu_u) / sigma_u
p_val_u = 2.0 * (1.0 - normal_cdf(abs(z_u)))

# --- E. 10,000-Iteration Monte Carlo Bootstrap ---
rng = np.random.default_rng(42)
boot_p_var = rng.binomial(n_var, p_var, size=10000) / n_var
boot_p_ctrl = rng.binomial(n_ctrl, p_ctrl, size=10000) / n_ctrl
boot_conv_lift = (boot_p_var - boot_p_ctrl) * 100.0
lift_ci = np.percentile(boot_conv_lift, [2.5, 97.5])

# --- F. Chi-Square Independence Test (Device Stratification, df=2) ---
obs = pd.crosstab(df['device'], df['converted']).to_numpy()
expected = (obs.sum(axis=1, keepdims=True) @ obs.sum(axis=0, keepdims=True)) / obs.sum()
chi2_device = np.sum((obs - expected)**2 / expected)
p_val_chi2 = math.exp(-0.5 * chi2_device)  # Exact survival function for df=2

# ==============================================================================
# 3. EXECUTIVE STATISTICAL DIAGNOSTIC REPORT
# ==============================================================================
print("=================================================================")
print(" STATISTICAL EXPERIMENTATION & CAUSAL INFERENCE ENGINE — SUMMARY")
print("=================================================================")
print(f"[1] SRM Integrity Check:   Control={n_ctrl:,} | Variant B={n_var:,} | χ²={srm_chi2:.4f} (p={p_val_srm:.4f}) -> {'PASS' if p_val_srm > 0.01 else 'FAIL'}")
print(f"[2] Conversion Rate Lift:  Control={p_ctrl:.2%} vs Variant B={p_var:.2%} (+{(p_var-p_ctrl)*100:.2f} pp)")
print(f"    Two-Proportion Z-Test: Z={z_conv:.4f} | p={p_val_conv:.6f} | Cohen's h={cohens_h:.4f}")
print(f"    Bootstrap 95% CI:      [{lift_ci[0]:.2f} pp, {lift_ci[1]:.2f} pp]")
print(f"[3] CUPED Session Lift:    +{np.mean(cuped_var)-np.mean(cuped_ctrl):.2f}s | Variance Reduced by {var_reduction_pct:.1f}% | Z={z_cuped:.4f} (p={p_val_cuped:.2e})")
print(f"[4] Payer Revenue (Skew):  Median Control=${statistics.median(payers_ctrl):.2f} vs Variant B=${statistics.median(payers_var):.2f}")
print(f"    Mann-Whitney U Test:   U={u_stat:,.0f} | Z={z_u:.4f} | p={p_val_u:.6f}")
print(f"[5] Device Independence:   χ²={chi2_device:.4f} (df=2) | p={p_val_chi2:.6e}")

# ==============================================================================
# 4. 4-PANEL DARK EXECUTIVE DASHBOARD (#0B0F19 / #111827 / #2E3D5C)
# ==============================================================================
BG_COLOR = '#0B0F19'
CARD_COLOR = '#111827'
BORDER_COLOR = '#2E3D5C'
TEXT_PRIMARY = '#F9FAFB'
TEXT_MUTED = '#9CA3AF'
ACCENT_CTRL = '#60A5FA'
ACCENT_VAR = '#34D399'
ACCENT_GOLD = '#FBBF24'

plt.rcParams.update({
    'figure.facecolor': BG_COLOR,
    'axes.facecolor': CARD_COLOR,
    'axes.edgecolor': BORDER_COLOR,
    'axes.linewidth': 1.2,
    'text.color': TEXT_PRIMARY,
    'axes.labelcolor': TEXT_MUTED,
    'xtick.color': TEXT_MUTED,
    'ytick.color': TEXT_MUTED,
    'font.sans-serif': 'DejaVu Sans',
    'axes.titlesize': 12,
    'axes.titleweight': 'bold'
})

fig, axes = plt.subplots(2, 2, figsize=(15, 10))
fig.subplots_adjust(left=0.06, right=0.96, top=0.85, bottom=0.08, wspace=0.22, hspace=0.32)

fig.text(0.06, 0.94, "STATISTICAL EXPERIMENTATION & CAUSAL INFERENCE ENGINE", fontsize=17, fontweight='bold', color=TEXT_PRIMARY)
fig.text(0.06, 0.90,
         f"N = {n_users:,}  |  SRM p = {p_val_srm:.3f} (PASS)  |  Conv Lift = +{(p_var-p_ctrl)*100:.2f} pp (p = {p_val_conv:.4f})  |  CUPED Var Reduction = {var_reduction_pct:.1f}%",
         fontsize=11, color=ACCENT_VAR)

# Panel 1: Conversion Rate & 95% Wilson Confidence Intervals
ax1 = axes[0, 0]
groups = ['Control', 'Variant B']
rates = [p_ctrl * 100, p_var * 100]
yerr_low = [rates[0] - ci_ctrl[0] * 100, rates[1] - ci_var[0] * 100]
yerr_high = [ci_ctrl[1] * 100 - rates[0], ci_var[1] * 100 - rates[1]]

bars = ax1.bar(groups, rates, width=0.42, color=[ACCENT_CTRL, ACCENT_VAR], alpha=0.9)
ax1.errorbar(groups, rates, yerr=[yerr_low, yerr_high], fmt='none', ecolor=TEXT_PRIMARY, elinewidth=2, capsize=6)
ax1.set_ylim(0, max(rates) * 1.35)
ax1.set_ylabel("Conversion Rate (%)")
ax1.set_title("1. Conversion Rate & 95% Wilson Score CI", loc='left', pad=12, color=TEXT_PRIMARY)
ax1.grid(axis='y', linestyle='--', alpha=0.15, color=TEXT_MUTED)
for bar, rate in zip(bars, rates):
    ax1.text(bar.get_x() + bar.get_width()/2, rate + 1.2, f"{rate:.2f}%", ha='center', fontweight='bold', color=TEXT_PRIMARY, fontsize=11)

# Panel 2: Monte Carlo Bootstrap Distribution of Lift (10,000 Iterations)
ax2 = axes[0, 1]
sns.histplot(boot_conv_lift, bins=45, kde=True, color=ACCENT_VAR, ax=ax2, edgecolor=BG_COLOR, alpha=0.65)
ax2.axvline(0, color='#EF4444', linestyle='--', linewidth=1.8, label='Null Hypothesis (0.0 pp)')
ax2.axvline(lift_ci[0], color=ACCENT_GOLD, linestyle=':', linewidth=1.8, label=f'95% CI [{lift_ci[0]:.2f}, {lift_ci[1]:.2f} pp]')
ax2.axvline(lift_ci[1], color=ACCENT_GOLD, linestyle=':', linewidth=1.8)
ax2.set_xlabel("Absolute Conversion Lift (Percentage Points)")
ax2.set_ylabel("Monte Carlo Frequency (10k Draws)")
ax2.set_title("2. Bootstrap Sampling Distribution of Lift", loc='left', pad=12, color=TEXT_PRIMARY)
ax2.legend(frameon=True, facecolor=BG_COLOR, edgecolor=BORDER_COLOR, fontsize=9)
ax2.grid(axis='y', linestyle='--', alpha=0.15, color=TEXT_MUTED)

# Panel 3: CUPED Variance Reduction Comparison (Raw vs. Adjusted SE)
ax3 = axes[1, 0]
metrics = ['Raw Session\nDuration', 'CUPED-Adjusted\nDuration']
ses = [se_raw, se_cuped]
b_cuped = ax3.bar(metrics, ses, width=0.42, color=[ACCENT_CTRL, ACCENT_GOLD], alpha=0.9)
ax3.set_ylim(0, max(ses) * 1.35)
ax3.set_ylabel("Standard Error of Treatment Effect (sec)")
ax3.set_title(f"3. CUPED Causal Variance Reduction (-{var_reduction_pct:.1f}% Var)", loc='left', pad=12, color=TEXT_PRIMARY)
ax3.grid(axis='y', linestyle='--', alpha=0.15, color=TEXT_MUTED)
for bar, val in zip(b_cuped, ses):
    ax3.text(bar.get_x() + bar.get_width()/2, val + 0.08, f"SE = {val:.3f}s", ha='center', fontweight='bold', color=TEXT_PRIMARY, fontsize=10)

# Panel 4: Stratified Conversion Lift by Device (Simpson's Paradox Check)
ax4 = axes[1, 1]
seg = df.groupby(['device', 'group'])['converted'].mean().unstack() * 100
seg = seg.loc[['Desktop', 'Mobile', 'Tablet']]
x = np.arange(len(seg.index))
w = 0.35

b1 = ax4.bar(x - w/2, seg['Control'], width=w, color=ACCENT_CTRL, label='Control')
b2 = ax4.bar(x + w/2, seg['Variant_B'], width=w, color=ACCENT_VAR, label='Variant B')
ax4.set_xticks(x)
ax4.set_xticklabels(seg.index)
ax4.set_ylim(0, seg.values.max() * 1.30)
ax4.set_ylabel("Conversion Rate (%)")
ax4.set_title(f"4. Device Stratification (χ² = {chi2_device:.2f}, p < 0.001)", loc='left', pad=12, color=TEXT_PRIMARY)
ax4.legend(frameon=True, facecolor=BG_COLOR, edgecolor=BORDER_COLOR, fontsize=9)
ax4.grid(axis='y', linestyle='--', alpha=0.15, color=TEXT_MUTED)
for bar_group in [b1, b2]:
    for bar in bar_group:
        h = bar.get_height()
        ax4.text(bar.get_x() + bar.get_width()/2, h + 0.4, f"{h:.1f}%", ha='center', fontsize=9, color=TEXT_PRIMARY)

plt.savefig('statistical_experimentation_dashboard.png', dpi=300, facecolor=BG_COLOR)
plt.show()
```
